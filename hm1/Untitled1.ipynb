```python
import numpy as np

url = (
    "https://raw.githubusercontent.com/DataTalksClub/"
    "machine-learning-zoomcamp/main/cohorts/2026/data/"
    "car_fuel_efficiency_2026.csv"
)

df = pd.read_csv(url)
df.head()
```

```python
asia = df[df["origin"] == "Asia"]

asia["fuel_efficiency_mpg"].max()
```

```python
median_before = df["horsepower"].median()
most_frequent = df["horsepower"].mode().iloc[0]

horsepower_filled = df["horsepower"].fillna(most_frequent)
median_after = horsepower_filled.median()

print("before:", median_before)
print("Most frequent:", most_frequent)
print("after:", median_after)
```

```python
X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print(w)
print(w.sum())
```
